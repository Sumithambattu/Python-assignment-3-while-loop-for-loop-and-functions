## While loop and control statement(else,break,continue,pass)

```python
# import random

# Generate a random number between 1 and 10
secret_number = random.randint(1, 10)
# Prompt the user
attempts = 3

print("Welcome to the Number Guessing Game!")
print("Guess a number between 1 and 10. You have 3 attempts.")

# implimenting the guessing logic
while attempts > 0:
  guess = int(input("Enter your guess: "))

   if guess < 1 or guess > 10:
    print("Invalid guess! Please enter a number between 1 and 10.")
       
  #control statements  
    continue

   attempts -= 1

   if guess > secret_number:
    print(f"Too high! {attempts} attempts left.")

  
  elif guess < secret_number:
    print(f"Too low! {attempts} attempts left.")

  
  else:
    print("Congratulations! You guessed the correct number!")
    # Control Statements: use break to exit on correct guess
    break


else:
  print(f"Better luck next time! The correct number was {secret_number}.")
```

## Funcion:BMI calculator

```python
def calculate_bmi(weight, height):
    """
    Calculates and returns the Body Mass Index (BMI).
    """
    bmi = weight / (height ** 2)
    return bmi

# PROMPT THE USER FOR INPUT
try:
    user_weight = float(input("Enter your weight in kg: "))
    user_height = float(input("Enter your height in meters : "))

    # CALCULATE BMI USING THE FUNCTION
    bmi_result = calculate_bmi(user_weight, user_height)

    # DISPLAY THE RESULT ROUNDED TO THE DECIMAL PLACES
    print(f"\nYour calculated BMI is: {bmi_result:.2f}")

except ValueError:
    print("Please enter valid numerical values for weight and height.")
except ZeroDivisionError:
    print("Height cannot be zero.")
    
    
```
